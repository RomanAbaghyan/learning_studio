# NoSQL Database Management Systems

![NoSQL joke](https://i.pinimg.com/originals/fd/e7/dd/fde7dde468b7e9d9f47d99a490aa05b7.jpg)

## What is NoSQL?

* Relational Database Management Systems (MySQL, PostgreSQL) have been considered as the one-size-fits-all solution for providing data storage and its retrieval.

* The increasing need for scalability and new application requirements have created new challenges for traditional RDBMS.

* NoSQL (**N**ot **O**nly **SQL**) is an umbrella term for all database management systems that don't follow the popular RDBMS principles and often relate to large data sets.

* NoSQL implies that more than one storage mechanism could be used based on the application needs.



## How is it different?

NoSQL systems typically don't have 

* SQL-like query statements.

* Fixed schema

* Expensive join operations

### Distributed, Fault-tolerant Architecture
NoSQL systems allow to store your database on multiple nodes and maintain high-speed performance

### Linear Scalability
Adding more nodes, gives us a consistent increase in performance

![architecture](https://i.stack.imgur.com/RXygq.png)


## Scalability and CAP theorem

Limitations of the distributed databases:

* Consistency - Every node always needs to see the same data value at any given moment

* Availability - The system should continue to operate even if nodes in a cluster crash or some of the hardware/software parts are down at any time.

* Partition Tolerance - The systems continues to operate in the presence of network partitions. 

### CAP Theorem

Any distributed database with shared data, can have at most two of
the three desirable properties **Consistency**, **Availability**, and **Partition Tolerance**.

![CAP diagram](https://static.wixstatic.com/media/933668_c8bd69bd93a642db824b4ebaedd45b6f~mv2.png/v1/fill/w_406,h_350,al_c,usm_0.66_1.00_0.01/933668_c8bd69bd93a642db824b4ebaedd45b6f~mv2.png)


## ACID vs BASE principle

ACID (used in RDBMS) and BASE (used in NoSQL systems) are transaction control models for databases.

### ACID

* Atomicity - the transaction must happen as an all or nothing

* Consistency - If we move some data from one place to another, we shouldn't get a report showing that the data is moved from some table without addition to another table. (ie. Database
should block all reports during atomic operations)

* Isolation - each part of the transaction occurs without knowledge of other parts

* Durability - once the transaction is done it should be permanent. (If
database crashes, all the done transactions must be restored from the
last backup points).

The main focus of ACID model is the integrity and consistency of data above any other considerations.

Losing the availability by temporarily blocking mechanisms is the only
way to ensure the reliability and accuracy of the information given by
the system.

### BASE 

* Basic Availability - the system can be temporarily inconsistent so that transactions are manageable and all information and
services are "Basically Available".

* Soft-State - some inaccuracies are allowed and data
may change during usage to reduce the amount of used resources for
the sake of availability.

* Eventual Consistency - at certain point when all the
service logic is executed, the system will turn into a consistent state

The main focus of BASE systems is the availability of services above
any other considerations. We should allow new data to be stored all
time even at the risk of being out of sync for a short period of time.

BASE systems tend to be simpler and faster than ACID as they don't
require handling of many events by locking and unlocking resources like
in case of ACID systems

## Advantages of NoSQL Databases

* Simple and Flexible Structure = Schema Free

* Based on Key-Value Pairs

* Some Store types include Column Store, Document Store, Graph Store, etc.

* Open Source NoSQL databases don't need expensive licensing fees
and can run on inexpensive hardware.

* Expansion is always cheaper and easier than relational databases.

* It depends on Horizontal scaling by distributing the load over more
nodes. On the contrary, Relational databases scale vertically by replac-
ing main host with a more powerful one.



## Disadvantages of NoSQL Databases

* They don't support the reliability features of relational databases
and ACID transaction control systems (Atomicity, Consistency, Isolation, Durability).

* They trade consistency for the sake of availability, performance and
scalability.

* Each type of NoSQL databases always has a limited number of applications.

* Incompatibility with the SQL queries and the need of a manual/proprietary
querying language which adds more complexity.

* They don't support Joins/Group By/ Order By/ ACID transactions.

## Types of NoSQL Databases

![Data Store Types NoSQL](https://docs.microsoft.com/en-us/dotnet/architecture/cloud-native/media/types-of-nosql-datastores.png)


### Key-Value Store

* Hash table is used to map each unique key to an item

* keys can be organized into logical group of keys ((ie. Identical keys
exist in different logical groups)

* All needed to access items in the database is the keys.

* Data can be stored in the form of a String, JSON, BLOB (Binary
Large Object), etc.

* They typically support CRUD (Create, Read, Update, Delete) opera
tions.

* Example software: [Amazon DynamoDB](https://aws.amazon.com/dynamodb/), [Apache Cassandra](http://cassandra.apache.org/) (supports SQL-like syntax).


### Wide-Column Store 

* Hybrid of RDBMS and Key-Value stores

* Data is stored in columns as opposed to being stored in rows as in
Relational database management systems.

* It consists of one or more column families that group certain columns
in the database and each key is used to identify and point to one
family.

* Each column includes tuples of names and values comma separated.

* Supports fast data aggregation

* Example: [Apache HBase](https://hbase.apache.org/)

![col_store](https://www.stechies.com/userfiles/images/row-colum.jpg)

### Document Store

* It is similar to key-value stores as they are schema-less and have
the same advantages/disadvantages.

* The value of each key stored in the database is a "document" with
many possible different encodings like XML, JSON, BSON (Binary
Encoded JSON).

* Documents themselves can contain many different key-value pairs, key-
array pairs, or nested documents.

* Documents can be indexed which leads to outperformance of
traditional file systems.

* Example: [MongoDB](https://www.mongodb.com/).

![doc_store](https://lennilobel.files.wordpress.com/2015/07/i4.png)

### Graph store

* A directed Graph structure is used to represent the data.
* The graph includes a set of **nodes, each represents an object**. Some
pairs of objects are connected by **links (edges) representing relations**
between objects.
* It is used typically in social networking applications.
* It allows developers to **focus on relations between objects** rather
than the objects themselves.
* They are powerful for **graph queries** like (shortest path, connected
components, etc.)
* Example: [Neo4J](https://neo4j.com/)

![graph](https://www.supinfo.com/articles/resources/214348/2138/0.png)



# Batch Processing

A single machine cannot handle the huge amount of data that's being generated every second.

We can either try to scale up or scale out:

* Monolithic: Twice cost to have double processing and memory does
not mean twice the performance

* Distributed: Many small cheap machines come together to act as a
single entity. Scales linearly as you add more nodes.

![](https://microsegment.io/img/scale-up-vs-scale-out.png)

All these machines need to be
coordinated by a single piece of
software to take care of

* Data partitioning

* Coordinate computing tasks

* Handle fault tolerance and
recovery

* Schedule tasks and allocate
resources․

Back in 2000, Google realized that traditional data management
approaches will not scale with Big data. They contributed two seminal
contributions.

* Google File System to manage distributed storage of data

* MapReduce programming model to abstract a distributed-parallel
data processing model


## Hadoop

* is an open source software framework 

* Clones the ideas of Google's File System, HDFS in addition to supporting MapReduce as the programming model

* Cluster management and scheduling of data
processing jobs

* Horizontal scaling as new nodes join the cluster

* Providea fault tolerance and transparent
rescheduling of failed jobs/tasks

* Manage resource allocation

### Hadoop Distributed File System (HDFS)

* Lowest level of data storage
* Partition at ingestion time
* Replicate for high-availability and fault
tolerance
* Supports serving several applications in
parallel on the same file partition
* Inspired by Google File System



### MapReduce 

* A simple abstraction for distributed data processing
* Invented by Google in 2004
* Two operations

    - **Map**: think of it as a transformation either 1:1 or 1:M that will be
applied to each element of your data

    - **Reduce**: think of it as a form of aggregating or compacting the data
M:1

* Hadoop provides the open source implementation of that model

* It was believed that all sorts of large-scale data processing can be
tackled by MapReduce, but this is not true as we will see.

#### MapReduce Overview

* You need to implement map,
reduce or both
* Map
  - Can be used to split elements,
  - Can be used to filter elements
* Reduce
  - To compute aggregates
  - To combine results

#### Word Count Example

* Partitioning
  - Distributes the key-space to the number of reducers available.
    - Number of reducers need not be the same as number of mappers
    - Assume two reducers

  - Makes sure that same keys go to the same reducer
  - Usually implemented via a hash function

* Shuffling
  - Preparing partitioned keys for the reduce step

* Sorting
  - Transforms (k,v1), (k,v2), (k,v3) $\to$ (k, {fv1,v2,v3})

### Yet Another Resource Negotiator (YARN) 

* Normally, a cluster runs several jobs

* Such jobs compete for cluster resources: CPU, memory, disk, network

* YARN is in charge of resource allocation. It has different scheduling policies:

  - FIFO - if there is one submitted job running, it uses all the resources of the cluster and another job will begin only after the first one finishes. This scheduler is rarely used, because there will be huge waiting times. 

  - Capacity - contains multiple queues with different jobs, each queue has some portion of the overall resources. Higher-priority jobs go to Queue 1. For jobs within the same queue, FIFO is typically used. Administrators can configure queues.
  - Fair - all jobs get equal share of resources

### Limitations of MapReduce

* Usability limitations
  - No support for joins
  - No support for iterative processing
  - No data or process sharing (e.g. many jobs may have the same logic for mappers, but we have to do the same process many times)
  - No utilization of indices

* Processing limitations
  - Intermediate steps are persisted to HDFS
  - Lack of Efficient/Compact data storage formats

## Apache Spark

* A general-purpose, fast, large-scale data
processing engine
* Started in AMPLab @ UC Berkley, now Data
Bricks
* Written in Scala
* Why would we need a new generation?
  - Limitations of MapReduce (Hadoop)
    - Materialized intermediate results
    - Just two abstractions Map and Reduce
    - Applications can only be written in Java
    - Poor support for real-time data processing
  - Exploit advancements in hardware
    - Memory is much cheaper
    - Multi-core is now a commodity
* Handle both batch and real-time processing
* Do not reinvent the wheel
  - Use HDFS for storage
  - Apache Mesos/YARN for execution
* Plays well with Hadoop
* Iterative processing

### Spark Abstractions

* Spark core abstraction is Resilient Distributed Dataset (RDD)
  - Resilient: fault tolerant and can be recomputed when recovering from
a failure
  - Distributed: processing takes place over several nodes in parallel, like
MapReduce
  - Dataset: initial data can come from files, memory, or created programmatically
  - Immutable: once created cannot be changed
  - Lineage: each RDD knows about its parents
* Spark applications are series of operations that transform input RDDs
into output RDDs or final values

### RDD Operations

* Two main types of RDD
operations
  - Transformations: result in a
new RDD
    - Can be chained
    - Forked
    - joined
  - Actions: return values, no
more RDDs
    - One action at the end of
each transformation chain




# Stream Processing

* Many applications require real-time processing of high-volume data
steams.

* They are pushing the limits of traditional data processing
infrastructures

## What is a Stream?

* Unbounded data
  - Conceptually infinite, ever growing set of data items/events
  - Requires continuous processing
* Notion of time
  - Time agnostic: stateless processing, count-based processing
  - Time-aware: event/ingestion/processing time 




## What is Stream Processing?

* Stream processing: is the processing of data as it is changing and
arriving continuously (data in motion).
  - In traditional (batch) processing, store data first, then process it.
  - In stream processing, process data on the fly



## Notion of Time in Stream Processing

* Typically, three notions of time can be used with stream processing
  - Ingestion time: the time at which a stream element arrives at the source
of the application
  - Processing time: the time at which an operator in the processing
pipeline reads the stream element
  - Event time: The original time at which the data was generated

* Progress of time
  - For both ingestion and processing time, the progress of time is controlled
by stream processing engine
  - Event time is controlled by the source. This leads to special characteristics for event time processing
    - Late arrival
    - Out of order




## Stream Windowing

* Time windows
  - Sliding

  - Tumbling

* Data-driven windows
  - Session windows

![](https://softwaremill.com/images/uploads/2016/10/types-of-windows.a93b3f60.png)

* Triggered by
  - Event time, processing time,
count



There is a skewness between event time
and processing time, which can be handled with **watermarks**:

* A timestamp indicating that no
elements should arrive with an earlier
timestamp
* If an element arrives with an earlier
timestamp, it is considered late



## Open Source Stream Processing Systems

* Kafka 
  - Reliable handling of huge numbers of concurrent reads and writes
  - Can be used as data-source / data-sink for Storm, Samza, Flink,
Spark and many more systems
  - Fault tolerant: Messages are persisted on disk and replicated within
the cluster. Messages (reads and writes) can be repeated

* Storm
  - True streaming over distributed data
ow
  - Low level API: Programmers have to specify the logic of each vertex
in the 
ow graph
  - Full understanding and hard coding of all used operators is required
  - Enables very high throughput (single purpose programs with small
overhead)

* Samza
  - True streaming built on top of Apache Kafka and Hadoop YARN
  - State is first class citizen
  - Low level API

* Apache Spark

  - Spark implements a batch execution engine
  - The execution of a job graph is done in stages
  - Operator outputs are materialized in memory (or disk) until the consuming operator is ready to consume the materialized data
  - Spark uses Discretized Streams (D-Streams)
  - Streams are interpreted as a series of deterministic batch-processing
jobs (A continuous series of RDDs)
  - Micro batches have a fixed granularity (default 3 seconds)
  - All windows defined in queries must be multiples of this granularity


* Flink

  - Flink's runtime is a native streaming engine
  - Based on Nephele/PACTs
  - Queries are compiled to a program in the form of an operator DAG
  - Operator DAGs are compiled to job graphs
  - Job graphs are generic streaming programs
  - Flink implements "true streaming" (handles batches by streamizing)
  


## Programming with Streams

* Low-Level API
  - A data
ow program is represented as a directed graph, whose nodes
represent a computation and whose edges represent connections among
data
ow nodes.
  - A stream-processing system distributes a data
ow graph across multiple
machines and is responsible for managing the partitioning of data, the
network communication, as well as program recovery in case of machine
failure.
  - Data
ow programming offers programmers complete freedom to imple-
ment their business logic, but require them to have good knowledge of
the execution internals.

* Functional APIs
  - stream-processing frameworks such as Spark or Flink offer higher-level
functional APIs. These APIs are more declarative than low-level data
ow
programming by giving programmers the ability to specify data-stream
programs as transformations on data-streams.

* High-Level Declarative Languages.
  - In the past, several research projects in stream processing have proposed
and some of them offerd declarative SQL-like language for data stream
processing.
  - Declarativity has the disadvantage of limiting the opportunities for fine-
tuning the performance of applications. However, a declarative language
allows for automatic optimization and shifts the responsibility of optimization from the programmer to the system.
