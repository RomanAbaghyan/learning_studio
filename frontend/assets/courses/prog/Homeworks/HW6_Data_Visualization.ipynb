Այստեղ մեր խնդիրն է պարզել, թե նոր ստացված կետը որ կլասին է պատկանում՝ օգտագործելով այդ կետի կոորդինատները։ Օրինակ՝ եթե կետը ավելի աջ է գտնվում, ապա այն ամենայն հավանականությամբ կապույտ կլասին է համապատասխան։

Նախ պետք է գրել ծրագիր ```classify```, որը կստանա կետերի կորդինատներ և (ստորև ներկայացված ալգորիթմով) կհաշվի ու կվերադարձնի դրանց համապատասխան պիտակները (0, եթե այն պատկանում է կարմիր կլասին և 1՝ հակառակ դեպքում)։ 

**ԱԼԳՈՐԻԹՄ**


Նշանակենք պիտակավորման ենթակա կետը $\mathbf{x}$֊ով։

1. Հաշվել յուրաքանչյուր կլասի centroid-ը (այն կետը, որի կոորդինատները հավասար են կլասի կետերի կոորդինատների թվաբանական միջիններին)։ Centroid-ները կնշանակենք $\mathbf{n}$(target=0) և $\mathbf{p}(target=1)$:

2. Հաշվել յուրաքանչյուր $\mathbf{x}$ կետի Էվկլիդյան հեռավորությունը $\mathbf{n}$֊ից և $\mathbf{p}$֊ից։
Օրինակ՝
$$\| x - p \|=\sqrt{( x - p )^{T}( x - p )}$$


3. Ծրագիրը պետք է վերադարձնի 1 պիտակը, եթե $\mathbf{x}$֊ը ավելի մոտ է $\mathbf{p}$-ին քան $\mathbf{n}$ և 0 հ. դ․։ Այսինքն` 1 պիտակը կլինի այն դեպքում, երբ
$$\mathbf{\| x - n \| \ge \| x - p \|} \\
\mathbf{\sqrt{( x - n )^{T}( x - n )} \ge \sqrt{( x - p )^{T}( x - p )}} \\
\mathbf{( x - n ) \cdot( x - n ) \ge ( x - p ) \cdot( x - p )} \\
\mathbf{x \cdot x -2 x \cdot n + n \cdot n \ge x \cdot x -2 x \cdot p + p \cdot p} \\
\mathbf{2 x \cdot p -2 x \cdot n \ge  p \cdot p - n \cdot n} \\
\mathbf{( p - n ) \cdot x \ge ( p - n ) \cdot( p + n ) / 2} \\
Նշանակենք՝\ 
\mathbf{w = p - n} ,\ t=\mathbf{( p - n ) \cdot( p + n ) / 2=\left(\| p \|^{2}-\| n \|^{2}\right) / 2} \\
\mathbf{w \cdot x} \ge t $$

5. Այսպիսով մեզ պետք է գտնել $\mathbf{w}$-ն և $t$-ն ու ըստ $\mathbf{w \cdot x} - t $ արտահայտության արժեքի պիտակավորել կետը (1, եթե $\mathbf{w \cdot x} - t >=0$ և 0 հ․դ․)

Պահանջվում է նորից տպել data֊ի կետերը, նշել կլասների միջնակետերը և գծել $\mathbf{w \cdot x}=t$ բաժանարար գիծը։
